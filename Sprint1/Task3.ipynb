```
using System;

public class Project
{
    public int ProjectId { get; set; }
    public string ProjectName { get; set; }
    public string Status { get; set; }

    public Project(int projectId, string projectName, string status)
    {
        ProjectId = projectId;
        ProjectName = projectName;
        Status = status;
    }

    public virtual void UpdateStatus(string newStatus)
    {
        Status = newStatus;
        Console.WriteLine($"Статус проекта обновлен на: {Status}");
    }

    public virtual string GetProjectDetails()
    {
        return $"ID: {ProjectId}, Название: {ProjectName}, Статус: {Status}";
    }

    public virtual void AssignTeamMember(string memberName)
    {
        Console.WriteLine($"Назначен участник: {memberName}");
    }
}

public class ProductProject : Project
{
    public DateTime ExpectedCompletionDate { get; set; }

    public ProductProject(int projectId, string projectName, string status, DateTime expectedCompletionDate)
        : base(projectId, projectName, status)
    {
        ExpectedCompletionDate = expectedCompletionDate;
    }

    public override void UpdateStatus(string newStatus)
    {
        base.UpdateStatus(newStatus);
        Console.WriteLine($"Ожидаемая дата завершения: {ExpectedCompletionDate:dd.MM.yyyy}");
    }
}

public class ResearchProject : Project
{
    public decimal FundingAmount { get; set; }

    public ResearchProject(int projectId, string projectName, string status, decimal fundingAmount)
        : base(projectId, projectName, status)
    {
        FundingAmount = fundingAmount;
    }

    public override void AssignTeamMember(string memberName)
    {
        base.AssignTeamMember(memberName);
        Console.WriteLine($"Требуемые навыки: Аналитика, Исследования, Научный подход");
        Console.WriteLine($"Объем финансирования: {FundingAmount:C}");
    }
}

public class InfrastructureProject : Project
{
    public int ExecutionPeriod { get; set; }

    public InfrastructureProject(int projectId, string projectName, string status, int executionPeriod)
        : base(projectId, projectName, status)
    {
        ExecutionPeriod = executionPeriod;
    }

    public override string GetProjectDetails()
    {
        return base.GetProjectDetails() + $", Срок выполнения: {ExecutionPeriod} мес.";
    }
}

class Program
{
    static void Main(string[] args)
    {
        ProductProject product = new ProductProject(101, "Разработка мобильного приложения", "В разработке", new DateTime(2026, 12, 28));
        ResearchProject research = new ResearchProject(202, "Исследование ИИ в медицине", "Исследование", 1500000m);
        InfrastructureProject infrastructure = new InfrastructureProject(303, "Строительство дата-центра", "Планирование", 18);

        Project[] projects = { product, research, infrastructure };

        foreach (var proj in projects)
        {
            Console.WriteLine(proj.GetProjectDetails());
        }

        Console.WriteLine();

        product.UpdateStatus("Готов к релизу");
        research.AssignTeamMember("Мария Воробьева");
        infrastructure.UpdateStatus("В строительстве");

        Console.WriteLine($"\nПроект '{product.ProjectName}' связан с исследованием '{research.ProjectName}'");
        Console.WriteLine($"Общее финансирование исследования: {research.FundingAmount:C}");
    }
}
```
